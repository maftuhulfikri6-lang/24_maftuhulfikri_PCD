```python
# Membuat file input.txt berisi angka 1 sampai 10 secara otomatis
with open('input.txt', 'w') as f:
    for i in range(1, 11):
        f.write(f"{i}\n")
print("File input.txt berhasil dibuat!")
```

```python
# Membaca file input.txt dan memproses angkanya
with open('input.txt', 'r') as f:
    for line in f:
        angka = int(line.strip()) # Mengubah teks baris menjadi angka bulat (integer)

        # Pengecekan ganjil atau genap
        if angka % 2 != 0:  # Jika ganjil
            hasil = angka * 2
            print(f"Angka Ganjil {angka} x 2 = {hasil}")
        else:               # Jika genap
            print(f"Angka Genap {angka} = {angka}")
```
