### **Tugas 2 — Mengolah String**

Buat program untuk mengolah data nama dan kalimat dengan ketentuan:

- Simpan nama depan dan nama belakang dalam variabel.
- Gabungkan keduanya menjadi nama lengkap.
- Tampilkan sebuah kalimat menggunakan `upper()`, `lower()`, `capitalize()`, dan `title()`.
- Hitung jumlah kemunculan huruf `a` menggunakan `count()`.
- Tampilkan hasil setiap operasi.

Output yang diharapkan: nama lengkap, perubahan format kalimat, dan jumlah karakter.



````python
# 1. Simpan nama depan dan nama belakang dalam variabel
nama_depan = "Maftuhul"
nama_belakang = "Fikri"

# 2. Gabungkan keduanya menjadi nama lengkap
nama_lengkap = nama_depan + " " + nama_belakang
print("Nama Lengkap :", nama_lengkap)

print("-" * 40)

# 3. Tampilkan sebuah kalimat menggunakan upper(), lower(), capitalize(), dan title()
kalimat = "belajar pemrograman python itu sangat menyenangkan"
print("Kalimat Asli     :", kalimat)
print("Upper()          :", kalimat.upper())
print("Lower()          :", kalimat.lower())
print("Capitalize()     :", kalimat.capitalize())
print("Title()          :", kalimat.title())

print("-" * 40)

# 4. Hitung jumlah kemunculan huruf 'a' menggunakan count()
jumlah_a = kalimat.count('a')
print(f"Jumlah huruf 'a' : {jumlah_a}")
````

### **Tugas 3 — Operasi Matematika**

Buat program kalkulator sederhana dengan ketentuan:

- Minta pengguna memasukkan dua angka.
- Konversikan input menjadi `int` atau `float`.
- Hitung penjumlahan, pengurangan, perkalian, pembagian, dan pangkat.
- Tampilkan hasil setiap operasi.
- Tambahkan perhitungan luas persegi panjang berdasarkan input panjang dan lebar.

Output yang diharapkan: hasil kalkulasi berdasarkan angka yang dimasukkan pengguna.



````python
# 1. Minta pengguna memasukkan dua angka dan konversikan ke float/int
print("--- KALKULATOR SEDERHANA ---")
angka1 = float(input("Masukkan angka pertama: "))
angka2 = float(input("Masukkan angka kedua: "))

print("-" * 30)

# 2. Hitung penjumlahan, pengurangan, perkalian, pembagian, dan pangkat
penjumlahan = angka1 + angka2
pengurangan = angka1 - angka2
perkalian = angka1 * angka2
pembagian = angka1 / angka2 if angka2 != 0 else "Tidak terhingga (pembagian dengan nol)"
pangkat = angka1 ** angka2

# 3. Tampilkan hasil setiap operasi kalkulator
print(f"Hasil Penjumlahan ({angka1} + {angka2}) : {penjumlahan}")
print(f"Hasil Pengurangan ({angka1} - {angka2}) : {pengurangan}")
print(f"Hasil Perkalian   ({angka1} * {angka2}) : {perkalian}")
print(f"Hasil Pembagian   ({angka1} / {angka2}) : {pembagian}")
print(f"Hasil Pangkat     ({angka1} ** {angka2}): {pangkat}")

print("=" * 30)

# 4. Perhitungan luas persegi panjang berdasarkan input panjang dan lebar
print("--- PERHITUNGAN LUAS PERSEGI PANJANG ---")
panjang = float(input("Masukkan panjang : "))
lebar = float(input("Masukkan lebar   : "))

luas = panjang * lebar
print(f"Luas Persegi Panjang ({panjang} * {lebar}) : {luas}")
````

### **Tugas 4 — Perulangan `for` dan `while`**

Buat program untuk menampilkan data mahasiswa dengan ketentuan:

- Buat daftar minimal lima nama mahasiswa.
- Tampilkan semua nama menggunakan `for`.
- Tampilkan angka 1 sampai 10 menggunakan `range()`.
- Tampilkan kembali daftar nama menggunakan `while` dan `len()`.
- Buat perulangan `while` untuk menampilkan bilangan genap dari 2 sampai 20.

Output yang diharapkan: daftar nama dan deret angka yang tampil sesuai ketentuan.


````python
# 1. Buat daftar minimal lima nama mahasiswa
mahasiswa = ["Maftuhul Fikri", "Ahmad", "Budi", "Siti", "Dewi"]

print("--- 1. Menampilkan Daftar Nama Menggunakan for ---")
# 2. Tampilkan semua nama menggunakan for
for nama in mahasiswa:
    print(f"- {nama}")

print("\n--- 2. Menampilkan Angka 1 sampai 10 Menggunakan range() ---")
# 3. Tampilkan angka 1 sampai 10 menggunakan range()
for i in range(1, 11):
    print(i, end=" ")
print() # Baris baru

print("\n--- 3. Menampilkan Daftar Nama Menggunakan while dan len() ---")
# 4. Tampilkan kembali daftar nama menggunakan while dan len()
i = 0
while i < len(mahasiswa):
    print(f"Indeks {i}: {mahasiswa[i]}")
    i += 1

print("\n--- 4. Bilangan Genap dari 2 sampai 20 Menggunakan while ---")
# 5. Buat perulangan while untuk menampilkan bilangan genap dari 2 sampai 20
genap = 2
while genap <= 20:
    print(genap, end=" ")
    genap += 2
print()
````

**Tugas 5 — Fungsi dan Waktu**

Buat program pencatatan aktivitas belajar dengan ketentuan:

- Impor modul `datetime`.
- Buat fungsi bernama `catat_waktu()`.
- Tampilkan pesan `Aktivitas selesai`.
- Tampilkan waktu saat fungsi dijalankan menggunakan `datetime.datetime.now()`.
- Panggil fungsi sebelum dan sesudah perulangan.
- Gunakan perulangan `for` untuk menampilkan proses pengerjaan lima soal.

Output yang diharapkan: pesan aktivitas, waktu pelaksanaan, dan proses pengerjaan soal.




````python
import datetime

# 1. Buat fungsi bernama catat_waktu()
def catat_waktu():
    print("Aktivitas selesai")
    # Tampilkan waktu saat fungsi dijalankan menggunakan datetime.datetime.now()
    waktu_sekarang = datetime.datetime.now()
    print(f"Waktu pelaksanaan: {waktu_sekarang}")

# --- Panggil fungsi sebelum perulangan ---
print("=== MULAI AKTIVITAS BELAJAR ===")
catat_waktu()

print("\n--- Proses Pengerjaan Soal ---")
# 2. Gunakan perulangan for untuk menampilkan proses pengerjaan lima soal
for i in range(1, 6):
    print(f"Sedang mengerjakan Soal ke-{i}...")

print("\n=== SELESAI AKTIVITAS BELAJAR ===")
# --- Panggil fungsi sesudah perulangan ---
catat_waktu()
````

## D. Tugas proyek akhir — Sistem Rekap Nilai Mahasiswa

Sebuah program digunakan untuk mencatat data mahasiswa, menerima nilai, menghitung rata-rata, dan menampilkan waktu penyelesaian proses.

Spesifikasi program:

1. Tampilkan judul program menggunakan `print()` dan garis pemisah.
2. Minta input nama dan NIM mahasiswa.
3. Minta pengguna memasukkan nilai tiga mata kuliah.
4. Hitung total dan rata-rata nilai menggunakan operasi matematika.
5. Tampilkan nama mahasiswa dalam format huruf kapital menggunakan `upper()`.
6. Gunakan `for` untuk menampilkan seluruh nilai.
7. Gunakan `while` untuk menampilkan nomor urut mata kuliah.
8. Buat fungsi `tampilkan_hasil()` untuk menampilkan total dan rata-rata.
9. Buat fungsi `catat_waktu()` untuk menampilkan waktu saat program menyelesaikan proses.

### Contoh gambaran output proyek akhir

```
================================
       REKAP NILAI MAHASISWA
================================
Nama       : ANDI PRATAMA
NIM        : 123456
--------------------------------
Nilai Algoritma   : 80
Nilai Basis Data  : 85
Nilai Struktur Data: 90
--------------------------------
Total Nilai : 255
Rata-rata   : 85.0
--------------------------------
Status      : Rekap selesai
Waktu       : [waktu program dijalankan]

```


Data di atas hanya contoh tampilan. Mahasiswa harus membuat program yang menghasilkan nilai berdasarkan input pengguna.


````python
import datetime

# 8. Buat fungsi tampilkan_hasil() untuk menampilkan total dan rata-rata
def tampilkan_hasil(total, rata_rata):
    print("-" * 35)
    print(f"Total Nilai : {total}")
    print(f"Rata-rata   : {rata_rata}")
    print("-" * 35)

# 9. Buat fungsi catat_waktu() untuk menampilkan waktu saat program menyelesaikan proses
def catat_waktu():
    waktu_sekarang = datetime.datetime.now()
    print("Status      : Rekap selesai")
    print(f"Waktu       : {waktu_sekarang}")

# 1. Tampilkan judul program menggunakan print() dan garis pemisah
print("===================================")
print("      REKAP NILAI MAHASISWA        ")
print("===================================")

# 2. Minta input nama dan NIM mahasiswa
nama_input = input("Masukkan Nama Mahasiswa : ")
nim = input("Masukkan NIM Mahasiswa  : ")

# 5. Tampilkan nama mahasiswa dalam format huruf kapital menggunakan upper()
nama = nama_input.upper()

print("-" * 35)

# 3. Minta pengguna memasukkan nilai tiga mata kuliah
nilai_ Algoritma = float(input("Nilai Algoritma      : "))
nilai_BasisData = float(input("Nilai Basis Data     : "))
nilai_StrukturData = float(input("Nilai Struktur Data  : "))

# Simpan nilai ke dalam list untuk memudahkan perulangan
daftar_nilai = [
    ("Algoritma", nilai_ Algoritma),
    ("Basis Data", nilai_BasisData),
    ("Struktur Data", nilai_StrukturData)
]

print("-" * 35)
print(f"Nama    : {nama}")
print(f"NIM     : {nim}")
print("-" * 35)

# 6 & 7. Gunakan for dan while untuk menampilkan seluruh nilai beserta nomor urutnya
i = 0
while i < len(daftar_nilai):
    matkul, nilai = daftar_nilai[i]
    # Nomor urut menggunakan perulangan while (indeks + 1)
    print(f"{i+1}. Nilai {matkul:<13}: {nilai}")
    i += 1

# 4. Hitung total dan rata-rata nilai menggunakan operasi matematika
total_nilai = nilai_ Algoritma + nilai_BasisData + nilai_StrukturData
rata_rata_nilai = total_nilai / len(daftar_nilai)

# Panggil fungsi untuk menampilkan total dan rata-rata
tampilkan_hasil(total_nilai, rata_rata_nilai)

# Panggil fungsi pencatatan waktu selesai
catat_waktu()
print("===================================")
````
