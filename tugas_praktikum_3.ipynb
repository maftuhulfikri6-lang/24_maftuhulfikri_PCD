**Soal 1 (OpenCV & Matplotlib):**
*   Baca gambar `orange.jpg` menggunakan OpenCV, ubah ruang warnanya ke RGB, lalu tampilkan menggunakan Matplotlib dengan ukuran plot $6 \times 6$.

```python
import cv2
import matplotlib.pyplot as plt

# 1. Baca gambar 'orange.jpg' menggunakan OpenCV
img_bgr = cv2.imread('orange.jpg')

# 2. Ubah ruang warna dari BGR ke RGB
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# 3. Tampilkan menggunakan Matplotlib dengan ukuran plot 6 x 6
plt.figure(figsize=(6, 6))
plt.imshow(img_rgb)
plt.axis('off')  # Menghilangkan sumbu x dan y agar tampilan lebih bersih (opsional)
plt.show()
```

**Soal 2 (NumPy - Cropping):**
*   Lakukan pemotongan (cropping) gambar pada koordinat baris/y $200$ hingga $600$ dan kolom/x $400$ hingga $800$ menggunakan slicing array NumPy, lalu tampilkan hasilnya.



```python
# 1. Pemotongan (cropping) gambar menggunakan slicing NumPy
# Format slicing: [baris_y_mulai:baris_y_selesai, kolom_x_mulai:kolom_x_selesai]
cropped_img = img_rgb[200:600, 400:800]

# 2. Tampilkan hasil cropping menggunakan Matplotlib
plt.figure(figsize=(6, 6))
plt.imshow(cropped_img)
plt.axis('off')
plt.show()
```

**Soal 3 (Scikit-Image - Filtering):**
*   Konversi hasil potongan (crop) dari Soal 2 menjadi grayscale, lalu terapkan filter deteksi tepi Sobel (`filters.sobel`) dari Scikit-Image dan tampilkan hasilnya dengan peta warna (cmap) abu-abu (`gray`).

```python
from skimage import color, filters

# 1. Konversi hasil potongan (cropped_img) dari Soal 2 menjadi grayscale
gray_img = color.rgb2gray(cropped_img)

# 2. Terapkan filter deteksi tepi Sobel menggunakan filters.sobel dari Scikit-Image
sobel_img = filters.sobel(gray_img)

# 3. Tampilkan hasilnya dengan peta warna (cmap) abu-abu ('gray')
plt.figure(figsize=(6, 6))
plt.imshow(sobel_img, cmap='gray')
plt.axis('off')
plt.show()
```
